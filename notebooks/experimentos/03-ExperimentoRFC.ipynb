# Importando bibliotecas

```python
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from numpy import random
```

# Importando dados

```python
# importando dados processados para experimento do modelo criados no notebook "02-Tratamento.ipynb"
treino_x = pd.read_csv('../../data/processed/treino_x.csv')
teste_x = pd.read_csv('../../data/processed/teste_x.csv')
treino_y = pd.read_csv('../../data/processed/treino_y.csv')
teste_y = pd.read_csv('../../data/processed/teste_y.csv')
```

# Experimento

```python
# configurando os parametros do modelo classificador
# n_estiumators seria a quantidade de arvores que teremos no modelo para prever
# random_state seria nossa SEED
modelo = RandomForestClassifier(n_estimators=100,
                                random_state=123143)

# ajustando modelo para os dados de treinamento que temos
# treino_x = features do treino
# treino_y = target do treino
modelo.fit(treino_x, treino_y)

# obter o resultado do treino do modelo com um teste, e para calcular a pontuacao do teste (score), iremos calcular a acuracia do teste do modelo
# teste_x = features do teste
# teste_y = target do teste
acuracia = modelo.score(teste_x, teste_y)*100
print("O modelo testado teve resultado de acuracia: %.1f%%" % acuracia)
```
