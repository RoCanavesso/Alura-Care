## Valores ausentes

### Anotacoes

- Foram identificados 419 valores NaN nessa variavel
- Temos 569 registros totais na base e 419 registros NaN para variavel exames_33

```python
df.isna().sum()
```

```python
# removendo variavel com maior representacao de valores ausentes 
df = df.drop(columns={'exame_33'}).copy()
```

## Identificando variaveis numericas

```python
# selecionando variaveis numericas de df
df_num = df.select_dtypes(include='number')
df_num
```

## Heatmap de correlacao das variaveis


```python
# criando matriz de correlacao das variaveis numericas
matriz_corr = df_num.corr()

# criando grafico de correlacao de variaveis com matriz de correlacao calculada
fig_heatmap = px.imshow(
    matriz_corr,
    title='Heatmap de correlacao das variaveis',
    text_auto=True,
    height=1200
)

fig_heatmap.show()
```
