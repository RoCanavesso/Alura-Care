# Importando bibliotecas

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from numpy import random
```

# Importando dados

```python
df = pd.read_csv('../../data/raw/exames.csv', )
df
```

# Tratando dados

```python
# removendo variaveis nao interessantes
df = df.drop(columns={'exame_4', 'exame_29', 'exame_33'}).copy()
```

```python
df
```

# Separando dados para train test de modelo

```python
# definindo SEED para reprodutibilidade do train_test_split
SEED = 123143
random.seed(SEED)

# criando variaveis para train_test_split
# valores_exames = toda a base df exceto as variaveis 'id' e 'diagnostico'. Essa base representara nossas features
valores_exames = df.drop(columns={'id', 'diagnostico'}).copy()
# diagnostico = apenas a coluna 'diagnostico' da base de dados, ela sera nosso target
diagnostico = df[['diagnostico']].copy()

# separando os dados para train e test do modelo
treino_x, teste_x, treino_y, teste_y = train_test_split(valores_exames,
                                                        diagnostico, 
                                                        test_size=0.3)

# visualizando a separacao dos dados
display(treino_x)
display(teste_x)
display(treino_y)
display(teste_y)

# salvando bases treino_x, teste_x, treino_y e teste_y
treino_x.to_csv('../../data/processed/treino_x.csv', index=False)
teste_x.to_csv('../../data/processed/teste_x.csv', index=False)
treino_y.to_csv('../../data/processed/treino_y.csv', index=False)
teste_y.to_csv('../../data/processed/teste_y.csv', index=False)
```
